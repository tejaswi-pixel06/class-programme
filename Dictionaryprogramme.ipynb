```python
#4
my_dict = {
    101:"Amit",
    102:"Rahul",
    103:"Sneha"
}
print("Keys:")
for key in my_dict:
    print(key)
print("\nValues:")
for values in my_dict.values():
    print(values)
print("\nKeys and Values:")
for key,values in my_dict.items():
    print(key,values)
```

```python
#12
student={
    "Amit":58,
    "Sneha":86,
    "Tejaswi":70,
    "Rahul":42,
    "Priya":94,
}
highest_student = max(student,key= student.get)
print("Student with highest marks:", highest_student)
print("Highest marks:", student[highest_student])
```

```python
#13
number = [10,30,40,10,50,40,60,60,30,80,90,]
frequency = {}
for num in number :
    if num in frequency :
        frequency[num] = frequency[num]+ 1
    else:
        frequency[num] = 1

print("Frequency of elemenys:",frequency)
```

```python
#14
my_dict = {
    "amit":34,
    "priya":35,
    "rahul":36,
    "sneha":55
}
sorted_dict = dict(sorted(my_dict.items(),key=lambda item : item[1]))
print("sorted dictionary:",sorted_dict)

sorted_dict= dict(sorted(my_dict.items(),key=lambda item: item[1],
                        reverse =True))
print(sorted_dict)
```
