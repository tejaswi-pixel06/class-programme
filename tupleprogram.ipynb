```python
numbers=[(23,12,76),(34,89,45),(11,22,33),(67,9,77)]
new_list=[]
for tup in numbers:
    new_tup = tup[:-1] + (100,)
    new_list.append(new_tup)
print("original list:", numbers)
print("updated list:",new_list)
```

```python
numbers=(1,2,3,4,5)
max=max(numbers)
min=min(numbers)
print("max element :", max)
print("min element :", min)
```

```python
numbers= (1,2,3,4,5)
new_tuple = (numbers[-1],) + numbers[1:-1] + (numbers[0],)
print("original list:", numbers)
print("tuple after swapping:", new_tuple)
```
