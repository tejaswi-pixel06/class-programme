```python
#slicing mentioning the end and start 
import numpy as np

arr = np.array([1,2,3,4,5,6,7])
print(arr[1:5])
```

```python
#slicing mentioning only end 
import numpy as np

arr = np.array([1,2,3,4,5,6,7])
print(arr[:5])
```

```python
#slicing not mentioning the end and start 
import numpy as np
arr = np.array([1,2,3,4,5,6,7])
print(arr[:])
```

```python
#slicing mentioning only start 
import numpy as np
arr = np.array([1,2,3,4,5,6,7])
print(arr[4:])
```

```python
#slicing elements from the begining to index 4
import numpy as np
arr = np.array([1,2,3,4,5,6,7])
print(arr[:4])
```

```python
#negative slicing
import numpy as np
arr = np.array([1,2,3,4,5,6,7])
print(arr[-3:-1])
```

```python
#determing the step of slicing
import numpy as np
arr = np.array([1,2,3,4,5,6,7])
print(arr[1:5:2])
```

```python
#return every other element from the entire array
import numpy as np
arr = np.array([1,2,3,4,5,6,7])
print(arr[::2])
```
