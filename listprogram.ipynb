```python
numbers= []
for i in range(20): 
    num=int(input("enter the number")) 
    numbers.append(num) 
print("list:",numbers) 
#counting similar values and their index values 
count=0 
print("similar elements and index values:") 
for i in range(20): 
    if numbers.count(numbers[i])>1: 
        print(numbers[i],"->",end="") 
        for j in range(20): 
            if numbers[j]==numbers[i]: 
                print(j,end="") 
print()

#counting of even and odd no. 
even_count=0 
odd_count=0 
for num in numbers:
  if num%2==0: 
    even_count+=1 
  else: 
    odd_count+=1 
print("even count:",even_count) 
print("odd count:",odd_count)

#counting of postive no. and negative no.
postive_count = 0
negative_count = 0
for num in numbers:
    if num>0:
        postive_count+=1
    else:
        negative_count+=1
print("postive count:",postive_count)
print("negative count:",negative_count)
        
```

```python
#List with duplicate elements
my_list = [10, 20, 20, 30, 10, 40, 30]

# Method 1: Using set() (Order is NOT preserved)
unique_list = list(set(my_list))
print("Without duplicates (unordered):", unique_list)

# Method 2: Using dict.fromkeys() (Order IS preserved)
unique_ordered_list = list(dict.fromkeys(my_list))
print("Without duplicates (ordered):", unique_ordered_list)
#i didnt understood this thing (do it again)
```
